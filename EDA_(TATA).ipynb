```python
import pandas as pd

# Load your CSV
df = pd.read_csv("Delinquency_prediction_dataset.csv")

# Median imputation
df['Income'].fillna(df['Income'].median(), inplace=True)
df['Loan_Balance'].fillna(df['Loan_Balance'].median(), inplace=True)
df['Credit_Score'].fillna(df['Credit_Score'].median(), inplace=True)

# Optional: Save the cleaned file
df.to_csv("Cleaned_Delinquency_Dataset.csv", index=False)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the cleaned dataset
df = pd.read_csv("/content/Cleaned_Delinquency_Dataset.csv")

# STEP 1: Compute correlation matrix (only numeric columns)
correlation = df.select_dtypes(include='number').corr()
print("Top correlations with Delinquent_Account:")
print(correlation['Delinquent_Account'].sort_values(ascending=False))

# Optional: Save correlation matrix to file
correlation.to_csv("correlation_matrix.csv")

# STEP 2: Boxplot – Credit Utilization vs Delinquency
sns.boxplot(x='Delinquent_Account', y='Credit_Utilization', data=df)
plt.title("Credit Utilization vs Delinquency")
plt.show()

# STEP 3: Boxplot – Missed Payments vs Delinquency
sns.boxplot(x='Delinquent_Account', y='Missed_Payments', data=df)
plt.title("Missed Payments vs Delinquency")
plt.show()

# STEP 4: Boxplot – Debt-to-Income Ratio vs Delinquency
sns.boxplot(x='Delinquent_Account', y='Debt_to_Income_Ratio', data=df)
plt.title("Debt-to-Income Ratio vs Delinquency")
plt.show()

# STEP 5: Countplot – Employment Status vs Delinquency
sns.countplot(x='Employment_Status', hue='Delinquent_Account', data=df)
plt.title("Employment Status vs Delinquency")
plt.xticks(rotation=30)
plt.show()
```

```python
# EDA Report
# Correlation Heatmap

import seaborn as sns
import matplotlib.pyplot as plt

# Select only numeric columns
numeric_df = df.select_dtypes(include='number')

# Plot heatmap
plt.figure(figsize=(10, 8))  # Optional: Adjust size
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Matrix of Numeric Variables")
plt.show()
```
